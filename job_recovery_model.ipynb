```python
import pandas as pd
```

```python
data = {
    "Role" : [
        "Data Analyst",
        "Data Scientist",
        "Data Engineer",
        "Machine Learning Engineer",
        "AI Engineer",
        "Frontend Developer",
        "Backend Developer",
        "Full Stack Developer",
        "Software Engineer",
        "Mobile App Developer",
        "Blockchain Developer",
        "DevOps Engineer",
        "Cloud Engineer",
        "Network Engineer",
        "Cybersecurity Analyst",
        "UI/UX Designer",
        "Graphic Designer",
        "Businees Analyst",
        "QA Engineer",
        "Technical Support Engineer"
        ],

    "Skills" : [
        "Python|SQL|Excel|Power BI",
        "Python|Machine Learning|Statistics|Deep Learning",
        "Python|SQL|ETL|Big Data",
        "Python|Machine Learning|Deep Learning|Tensor Flow",
        "Python| Deep Learning|NLP|AI",
        "HTML|CSS|Javascript|React",
        "Python|Django|SQL|APIs",
        "HTML|CSS|Javascript|Python",
        "Java|DSA|OOP|System Designer",
        "Java|Kotlin|Android|APIs",
        "Blockchain|Solidity|Cryptography",
        "Docker|Kubernetes|CI/CD|Linux",
        "AWS|Azure|Cloud|Networking",
        "Networking|Routing|Linux",
        "Security|Networking|Ethical Hacking",
        "Figma|Wireframing|User Research",
        "Photoshop|Illustrator|Creativity",
        "Excel|SQL|Communication|Analysis",
        "Testing|Automation|Selenium",
        "Troubleshooting|Networking|Customer Support"
    ]
}
```

```python
len(data["Role"])
```

```python
len(data["Skills"])
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df["Skills"] = df["Skills"].apply(lambda x: x.split("|"))
```

```python
df["Skills"] = df["Skills"].apply(lambda x:[skill.lower() for skill in x])
```

```python
df.head()
```

```python
user_input = input("Enter your skills : ")
user_skills = [skill.strip().lower() for skill in user_input.split(",")]
print(user_skills)
```

```python
results = []
for index, row in df.iterrows():
    role = row["Role"]
    required_skills = row["Skills"]

    matched = [skill for skill in user_skills if skill in required_skills]

    score = (len(matched)/len(required_skills))*100

    missing = [skill for skill in required_skills if skill not in user_skills]

    results.append({
        "Role" : role,
        "Match %" : round(score,2),
        "Missing Skills" : missing
    })
    
```

```python
for r in results:
    print(r)
```

```python
results_df = pd.DataFrame(results)
results_df
```

```python
results_df = results_df.sort_values(by = "Match %", ascending = False)
```

```python
top_jobs = results_df.head(5)
top_jobs
```

```python
for index , row in top_jobs.iterrows():
    print("Role : ",row["Role"])
    print("Match % : ",row["Match %"])
    print("Missing Skills :", ",". join (row["Missing Skills"]))
    print("-" * 40)
```

```python
results_df = results_df[results_df["Match %"]>=30]
```

```python
top_jobs = results_df.head(5)
```

```python
top_jobs
```

```python
if len(top_jobs) == 0:
    print("No matching jobs found . Try adding more skills.")
else:
    print("Top Job Recommendation :\n")
    for index,row in top_jobs.iterrows():
        print("Role :", row["Role"])
        print("Match % :" , round(row["Match %"],1))
        print("Missing Skills :",",".join([skill.upper() for skill in row["Missing Skills"]]))
        print("-" * 40)
        
```

```python
best_job = top_jobs.iloc[0]
print("\nBest Match for You:")
print("Role:",best_job["Role"])
print("Match %",best_job["Match %"])
```
